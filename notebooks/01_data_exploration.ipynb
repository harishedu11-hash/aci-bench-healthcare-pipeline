```python
import pandas as pd

df = pd.read_csv("data/challenge_data/train.csv")

print(df.shape)
print(df.columns.tolist())
```

```python
df.info()
```

```python
df["dialogue_chars"] = df["dialogue"].str.len()
df["note_chars"] = df["note"].str.len()

print(df[["dialogue_chars", "note_chars"]].describe())
```

```python
print(df.loc[0, "dialogue"][:2000])
print("\n" + "=" * 80 + "\n")
print(df.loc[0, "note"])
```

```python
for i in range(3):
    print(f"\n{'='*80}\nENCOUNTER: {df.loc[i, 'encounter_id']}\n")
    print(df.loc[i, "note"][:1000])
```

```python
summary = {
    "rows": len(df),
    "columns": len(df.columns),
    "avg_dialogue_chars": round(df["dialogue_chars"].mean(), 2),
    "avg_note_chars": round(df["note_chars"].mean(), 2),
}

summary
```
